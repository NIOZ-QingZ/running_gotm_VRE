# Sensitivity of filter feeders to TEP excretion (`p_pueR2`)

Suspension feeders (`Y3`, and juveniles `Yy3`) filter TEP (`R2`) from the water.
Of the TEP they take up, a fraction `p_pueR2` is excreted again (as faeces, to `R6`);
the rest is assimilated. The OysterGrounds case uses `p_pueR2 = 0.90` (90 %).
This notebook compares that with 40 %.

`p_pueR2` is a namelist parameter in `FilterFeeder.nml` (`/FilterFeeder_parameters/`),
read by `src/BFM/Ben/FilterFeeder.F90` (`se_uR2 = suf*p_pueR2(y)`), so **no recompilation
is needed**: the current image runs it as is. The parameter has one value per filter-feeder
group (`Y3`, `Yy3`), so both are set.

`p_pueR3` (TEP inside *Phaeocystis* colonies, `R3`) is also 0.90 in the case; the
third scenario changes it as well. Remove that scenario if you only want `R2`.

Plots of the mean suspension-feeder biomass (`Y3c`) over one year:

* **(a)** annual mean `Y3c` per scenario, labelled with the % change from the 90 % run
* **(b)** daily `Y3c` through the year, one line per scenario

One year takes about 6 minutes per run; the runs are started in parallel.

```python
# Settings
# Each scenario: label -> list of FILE:[GROUP:]NAME=VALUE settings for gotm-run --set.
# The first scenario is the reference for the % change.
SCENARIOS = {
    "pueR2_0.90":       ["FilterFeeder.nml:p_pueR2=0.90,0.90"],    # case default
    "pueR2_0.40":       ["FilterFeeder.nml:p_pueR2=0.40,0.40"],
    "pueR2_R3_0.40":    ["FilterFeeder.nml:p_pueR2=0.40,0.40",
                         "FilterFeeder.nml:p_pueR3=0.40,0.40"],
}
START = "1972-01-01 00:00:00"
STOP = "1973-01-01 00:00:00"
YEAR = 1972            # year to average over
VAR = "Y3c"            # BFM variable in OG.nc, e.g. "jnetY3c" for net production
RESULTS_ROOT = "results/TEPexcretion"
MAX_PARALLEL = 6       # runs at the same time
```

```python
# GOTM runner
import os
import shutil
import subprocess
from concurrent.futures import ThreadPoolExecutor
from pathlib import Path


def run_gotm(label, start=START, stop=STOP, root=RESULTS_ROOT):
    """One GOTM-BFM run for SCENARIOS[label]; returns (label, returncode, run dir)."""
    results_dir = Path(root) / label
    if results_dir.exists():          # gotm-run needs a new or empty directory
        shutil.rmtree(str(results_dir))
    results_dir.parent.mkdir(parents=True, exist_ok=True)

    command = ["gotm-run", "--out", str(results_dir), "--start", start, "--stop", stop]
    for setting in SCENARIOS[label]:
        command += ["--set", setting]
    proc = subprocess.run(command, stdout=subprocess.PIPE, stderr=subprocess.STDOUT)
    return label, proc.returncode, str(results_dir)


print("Starting {} runs from {} to {} ...".format(len(SCENARIOS), START, STOP))
with ThreadPoolExecutor(max_workers=MAX_PARALLEL) as pool:
    runs = list(pool.map(run_gotm, SCENARIOS))

for label, returncode, run_dir in runs:
    status = "finished" if returncode == 0 else "FAILED (see {}/log.out)".format(run_dir)
    print("{:<16} {}".format(label, status))
```

```python
# Check that the model used the intended values (namelist echo in log.out)
import re

for label, returncode, run_dir in runs:
    with open(os.path.join(run_dir, "log.out"), errors="replace") as f:
        log = f.read()
    used = re.findall(r"P_PUER[23]\s*=[^A-Z]*", log, flags=re.IGNORECASE)
    print("{:<16} {}".format(label, " | ".join(" ".join(u.split()) for u in used) or "(not found in log.out)"))
```

```python
# Read the results: annual mean of VAR per run
import datetime
import netCDF4
import numpy as np


def read_year(path, var, year):
    """Dates and values of `var` within `year`, plus its units and long name."""
    with netCDF4.Dataset(path) as nc:
        t = nc.variables["time"]
        # num2date may return cftime dates, which matplotlib cannot plot:
        # convert them to standard Python datetimes
        dates = [datetime.datetime(d.year, d.month, d.day, d.hour, d.minute, d.second)
                 for d in netCDF4.num2date(t[:], t.units)]
        values = np.ma.filled(nc.variables[var][:].reshape(len(dates), -1)[:, 0], np.nan)
        units = getattr(nc.variables[var], "units", "")
        long_name = getattr(nc.variables[var], "long_name", var)
    keep = np.array([d.year == year for d in dates])
    return [d for d, k in zip(dates, keep) if k], values[keep], units, long_name


results = []    # (label, dates, values), in the order of SCENARIOS
for label, returncode, run_dir in runs:
    if returncode != 0:
        continue
    dates, values, units, long_name = read_year(os.path.join(run_dir, "OG.nc"), VAR, YEAR)
    results.append((label, dates, values))

labels = [r[0] for r in results]
means = np.array([np.nanmean(r[2]) for r in results])
reference = list(SCENARIOS)[0]
if reference not in labels:
    raise RuntimeError("reference run {} failed".format(reference))
ref = means[labels.index(reference)]
change = 100.0 * (means - ref) / ref

print("{:<16}  {:>12}  {:>8}".format("scenario", "mean " + VAR, "change"))
for l, m, c in zip(labels, means, change):
    print("{:<16}  {:>12.2f}  {:>+7.2f}%".format(l, m, c))
```

```python
# Plot
%matplotlib inline
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5))
colors = plt.cm.viridis(np.linspace(0, 0.9, len(results)))
x = np.arange(len(results))

ax1.bar(x, means, color=colors, width=0.6)
for xi, m, c in zip(x, means, change):
    ax1.annotate("{:+.1f}%".format(c), (xi, m), textcoords="offset points",
                 xytext=(0, 4), ha="center", fontsize=9)
ax1.set_xticks(x)
ax1.set_xticklabels(labels, fontsize=9)
ax1.set_xlabel("scenario (excreted fraction of TEP uptake)")
ax1.set_ylabel("annual mean {} [{}]".format(VAR, units))
ax1.set_title("(a) {}, mean over {}".format(long_name, YEAR), fontsize=10)
ax1.margins(y=0.12)

for (label, dates, values), col in zip(results, colors):
    ax2.plot(dates, values, color=col, label=label)
ax2.set_ylabel("{} [{}]".format(VAR, units))
ax2.set_title("(b) daily values in {}".format(YEAR), fontsize=10)
ax2.legend(title="scenario", fontsize=8, frameon=False)
fig.autofmt_xdate()
fig.tight_layout()

fig.savefig(os.path.join(RESULTS_ROOT, "TEPexcretion_{}_{}.png".format(VAR, YEAR)), dpi=150)
plt.show()
```
