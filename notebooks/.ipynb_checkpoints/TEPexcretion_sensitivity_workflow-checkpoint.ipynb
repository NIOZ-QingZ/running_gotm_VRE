# Sensitivity of filter feeders to TEP excretion (`p_pueR2`, `p_pueR3`) and `p_max`

Suspension feeders (`Y3`, and juveniles `Yy3`) take up two kinds of TEP:

* **R2**, free TEP, taken up together with diatoms (`P1`, `P5`) in proportion to the
  diatoms eaten. A fraction `p_pueR2` of it is excreted again (as faeces, to `R6`).
* **R3**, the carbon in *Phaeocystis* colonies outside the cells (the colony matrix).
  `Y3` eats *Phaeocystis* (`P6`; `p_puPI(1,6) = 1`) and takes up the colony's R3 with it.
  A fraction `p_pueR3` of it is excreted.

Both fractions are 0.90 in the OysterGrounds case. They also set the overall assimilation
efficiency of the food (`sunmc` in `FilterFeeder.F90`), which decides whether filtering is
worth its cost (`active`). A lower excretion fraction therefore gives more carbon from TEP,
and also more filtering of all food, mainly during the *Phaeocystis* bloom.

This notebook varies each fraction from 0.90 to 0.10 while the other stays at 0.90:

* `p_pueR2` = 0.90, 0.40, 0.20, 0.10 (with `p_pueR3` = 0.90)
* `p_pueR3` = 0.90, 0.40, 0.20, 0.10 (with `p_pueR2` = 0.90)

A third sweep varies `p_max` (case value 0.05) with both fractions at 0.90:

* `p_max` = 0.02, 0.05, 0.10, 0.50, 1.00

`p_max` has two roles in `FilterFeeder.F90`:

1. **Extraction cap** for phytoplankton, zooplankton and R6 detritus:
   `corr_max = p_max/LocalDelta` (about 14.4 d⁻¹ with `dt = 300 s`) limits uptake smoothly,
   by `corr_max/(uptake/stock + corr_max)`, against the stock in the bottom layer.
2. **Fixed factor on R2 (TEP) uptake**: the near-bed concentration factor of sinking R2
   (always above 1) is clipped at `p_max` before the actual uptake
   (`ruR2c = R2x_food_c*suf*Depth_Ben*min(s, p_max)`). With 0.05, the R2 eaten is 5 % of
   the R2 counted as food; with 1.0 there is no such reduction.

A `p_max` run therefore changes both at once.

All three are namelist parameters in `FilterFeeder.nml` (`/FilterFeeder_parameters/`), with one
value per filter-feeder group (`Y3`, `Yy3`) for `p_pueR2`/`p_pueR3` and one value for
`p_max`, so **no recompilation is needed**.
Do not use exactly 0 for `p_pueR2`: the model takes that as "no TEP uptake".

Outputs, over one year:

* **(a)** annual mean `Y3c` against the excretion fraction, one line for `p_pueR2` and
  one for `p_pueR3`, labelled with the % change from the reference (0.90/0.90, `p_max` 0.05)
* **(b)** annual mean `Y3c` against `p_max` (log scale)
* **(c)** daily `Y3c` through the year: `p_pueR2` runs solid, `p_pueR3` dashed, `p_max` dotted

One year takes about 6 minutes per run; the runs are started in parallel.

## NaaVRE workflow

```
[TEP scenarios] --> [run GOTM TEP] --> [plot TEP]
   param_scenarios: List   run_dirs: List   results_folder: String
```

1. **TEP scenarios**: the list of scenarios, items `"pueR2/pueR3"` or `"pueR2/pueR3/p_max"`.
2. **run GOTM TEP**: one model run per scenario, in parallel; outputs `run_dirs`, the run
   folders under `/tmp/data/` (the only folder shared between components).
3. **plot TEP**: reads `OG.nc` from each folder, makes the table and figure and saves them,
   with each run's `log.out` and `run_info.json`, to Cloud Storage, because NaaVRE clears
   `/tmp/data/` when the workflow finishes. It outputs that folder as `results_folder`.


````python
# TEP scenarios
# ---
# NaaVRE:
#  cell:
#   inputs: []
#   outputs:
#    - param_scenarios: List
# ...
# One item per model run: "pueR2/pueR3" or "pueR2/pueR3/p_max"
# (excreted fraction of TEP uptake for R2 and R3; FilterFeeder p_max, default 0.05).
# The case default is 0.90/0.90/0.05; it is used as the reference for the % change.
param_scenarios = [
    "0.90/0.90",                              # reference (case default)
    "0.40/0.90", "0.20/0.90", "0.10/0.90",    # p_pueR2 varied, p_pueR3 = 0.90
    "0.90/0.40", "0.90/0.20", "0.90/0.10",    # p_pueR3 varied, p_pueR2 = 0.90
    "0.90/0.90/0.02", "0.90/0.90/0.10",       # p_max varied (case value 0.05)
    "0.90/0.90/0.50", "0.90/0.90/1.00",
]
print("param_scenarios:", param_scenarios)
````

````python
# run GOTM TEP
# ---
# NaaVRE:
#  cell:
#   inputs:
#    - param_scenarios: List
#   outputs:
#    - run_dirs: List
#   params:
#    - param_year:
#       type: Integer
#       default_value: 1972
#    - param_var:
#       type: String
#       default_value: "Y3c"
#    - param_start:
#       type: String
#       default_value: "1972-01-01 00:00:00"
#    - param_stop:
#       type: String
#       default_value: "1973-01-01 00:00:00"
# ...


# Settings (do not containerize this cell)
# param_ values are the defaults of the workflow components; they can be changed when
# the workflow is submitted.
param_start = "1972-01-01 00:00:00"
param_stop = "1973-01-01 00:00:00"
param_results_root = "/tmp/data/TEPexcretion"   # shared between workflow components
param_year = 1972            # year to average over
param_var = "Y3c"            # BFM variable in OG.nc, e.g. "jnetY3c" for net production
param_cloud_folder = "/home/jovyan/Cloud Storage/naa-vre-user-data/TEPexcretion"   # kept after the workflow

# TEP scenarios: labels "pueR2_<x>_pueR3_<y>_pmax_<z>", one per model run.
# param_scenarios (List, from the TEP scenarios cell): items "pueR2/pueR3[/p_max]",
# e.g. "0.40/0.90" or "0.90/0.90/0.50", or lists [pueR2, pueR3, p_max];
# missing values are the case values (pueR3 0.90, p_max 0.05).
items = param_scenarios if isinstance(param_scenarios, (list, tuple)) else str(param_scenarios).split(",")
scenarios = []
for item in items:
    if isinstance(item, (list, tuple)):
        values = [float(v) for v in item]
    elif str(item).strip():
        values = [float(v) for v in str(item).split("/")]
    else:
        continue
    pueR2 = values[0]
    pueR3 = values[1] if len(values) > 1 else 0.90
    pmax = values[2] if len(values) > 2 else 0.05
    scenarios.append("pueR2_{:.2f}_pueR3_{:.2f}_pmax_{:g}".format(pueR2, pueR3, pmax))
print("scenarios:", scenarios)

# run GOTM TEP
# Output run_dirs (List of str): run folders under param_results_root.
import os
import re
import shutil
import subprocess
from concurrent.futures import ThreadPoolExecutor
import stat


def remove_tree(path):
    """Delete a file or folder. gotm-run copies the case from the read-only image, so
    its folders are read-only: add write permission (like chmod -R u+w) and retry."""
    def make_writable(func, p, exc_info):
        for q in (os.path.dirname(p), p):
            try:
                os.chmod(q, os.stat(q).st_mode | stat.S_IWUSR | stat.S_IRUSR | stat.S_IXUSR)
            except OSError:
                pass
        func(p)
    if os.path.isdir(path) and not os.path.islink(path):
        shutil.rmtree(path, onerror=make_writable)
    else:
        try:
            os.remove(path)
        except PermissionError:
            make_writable(os.remove, path, None)




def run_one(label):
    """One GOTM-BFM run for a label "pueR2_<x>_pueR3_<y>_pmax_<z>"; returns the run folder."""
    parts = label.split("_")
    pueR2, pueR3, pmax = float(parts[1]), float(parts[3]), float(parts[5])
    run_dir = os.path.join(param_results_root, label)
    if os.path.exists(run_dir):          # gotm-run needs a new or empty directory
        remove_tree(run_dir)
    os.makedirs(param_results_root, exist_ok=True)
    command = [
        "gotm-run",
        "--out", run_dir,
        "--start", param_start,
        "--stop", param_stop,
        # one value per filter-feeder group (Y3, Yy3)
        "--set", "FilterFeeder.nml:p_pueR2={0:g},{0:g}".format(pueR2),
        "--set", "FilterFeeder.nml:p_pueR3={0:g},{0:g}".format(pueR3),
        "--set", "FilterFeeder.nml:p_max={:g}".format(pmax),   # one value for all groups
    ]
    proc = subprocess.run(command, stdout=subprocess.PIPE, stderr=subprocess.STDOUT)
    status = "finished" if proc.returncode == 0 else "FAILED (see {}/log.out)".format(run_dir)
    # namelist echo in log.out: the values the model actually used
    used = ""
    log_file = os.path.join(run_dir, "log.out")
    if os.path.exists(log_file):
        with open(log_file, errors="replace") as f:
            used = " | ".join(" ".join(u.split()) for u in
                              re.findall(r"P_(?:PUER[23]|MAX)\s*=[^A-Z]*", f.read(), flags=re.IGNORECASE))
    print("{:<34} {}   {}".format(label, status, used or "(values not found in log.out)"), flush=True)
    if proc.returncode != 0:
        print(proc.stdout.decode(errors="replace")[-2000:])
    return run_dir


# Empty the run folder first, so files from an earlier submission are not mixed in
if os.path.isdir(param_results_root) and os.listdir(param_results_root):
    print("Emptying", param_results_root)
    for name in os.listdir(param_results_root):
        remove_tree(os.path.join(param_results_root, name))
os.makedirs(param_results_root, exist_ok=True)

print("Starting {} run(s) from {} to {} ...".format(len(scenarios), param_start, param_stop), flush=True)
with ThreadPoolExecutor(max_workers=max(1, len(scenarios))) as pool:
    run_dirs = list(pool.map(run_one, scenarios))
print("run_dirs:", run_dirs)
````

````python
# plot TEP
# ---
# NaaVRE:
#  cell:
#   inputs:
#    - run_dirs: List
#   outputs:
#    - results_folder: String
# ...

# plot TEP
# Input run_dirs (List of str): run folders from the run cell.
# Output results_folder (str): Cloud Storage folder with the figure and the table.
import datetime
import json
import os
import shutil

import matplotlib.pyplot as plt
import netCDF4
import numpy as np
import stat


def remove_tree(path):
    """Delete a file or folder. gotm-run copies the case from the read-only image, so
    its folders are read-only: add write permission (like chmod -R u+w) and retry."""
    def make_writable(func, p, exc_info):
        for q in (os.path.dirname(p), p):
            try:
                os.chmod(q, os.stat(q).st_mode | stat.S_IWUSR | stat.S_IRUSR | stat.S_IXUSR)
            except OSError:
                pass
        func(p)
    if os.path.isdir(path) and not os.path.islink(path):
        shutil.rmtree(path, onerror=make_writable)
    else:
        try:
            os.remove(path)
        except PermissionError:
            make_writable(os.remove, path, None)



# Flatten, in case run_dirs arrives as a list of lists
folders = []
for item in (run_dirs if isinstance(run_dirs, (list, tuple)) else [run_dirs]):
    folders.extend(item if isinstance(item, (list, tuple)) else [item])

year = int(param_year)
var = str(param_var)

results = []    # (pueR2, pueR3, pmax, dates, values)
units, long_name = "", var
for run_dir in folders:
    label = os.path.basename(os.path.normpath(run_dir))
    info_file = os.path.join(run_dir, "run_info.json")
    nc_file = os.path.join(run_dir, "OG.nc")
    if os.path.exists(info_file):
        with open(info_file) as f:
            if json.load(f).get("status") != "finished":
                print("skipped (run failed):", run_dir)
                continue
    if not os.path.exists(nc_file):
        print("skipped (no OG.nc):", run_dir)
        continue
    parts = label.split("_")              # pueR2_<x>_pueR3_<y>[_pmax_<z>]
    pueR2, pueR3 = float(parts[1]), float(parts[3])
    pmax = float(parts[5]) if len(parts) > 5 else 0.05
    with netCDF4.Dataset(nc_file) as nc:
        t = nc.variables["time"]
        # num2date may return cftime dates, which matplotlib cannot plot:
        # convert them to standard Python datetimes
        dates = [datetime.datetime(d.year, d.month, d.day, d.hour, d.minute, d.second)
                 for d in netCDF4.num2date(t[:], t.units)]
        values = np.ma.filled(nc.variables[var][:].reshape(len(dates), -1)[:, 0], np.nan)
        units = getattr(nc.variables[var], "units", "")
        long_name = getattr(nc.variables[var], "long_name", var)
    keep = np.array([d.year == year for d in dates])
    results.append((pueR2, pueR3, pmax, [d for d, k in zip(dates, keep) if k], values[keep]))
if not results:
    raise RuntimeError("no finished runs in {}".format(folders))
results.sort(key=lambda r: (-r[0], -r[1], r[2]))

# Annual mean and % change from the case default (0.90/0.90, p_max 0.05; or the first run)
REF = (0.90, 0.90, 0.05)
close = lambda a, b: abs(a - b) < 1e-6
labels = ["R2 {:.2f}, R3 {:.2f}, p_max {:g}".format(*r[:3]) for r in results]
means = np.array([np.nanmean(r[4]) for r in results])
i_ref = next((i for i, r in enumerate(results) if all(close(a, b) for a, b in zip(r[:3], REF))), 0)
ref = means[i_ref]
ref_r2, ref_r3, ref_pmax = results[i_ref][:3]
change = 100.0 * (means - ref) / ref

table = ["p_pueR2,p_pueR3,p_max,mean_{},change_percent".format(var)]
print("reference:", labels[i_ref])
print("{:>7}  {:>7}  {:>6}  {:>12}  {:>8}".format("p_pueR2", "p_pueR3", "p_max", "mean " + var, "change"))
for r, m, c in zip(results, means, change):
    print("{:>7.2f}  {:>7.2f}  {:>6g}  {:>12.2f}  {:>+7.2f}%".format(r[0], r[1], r[2], m, c))
    table.append("{:g},{:g},{:g},{:.6g},{:.4f}".format(r[0], r[1], r[2], m, c))

# Sweeps: one parameter varied, the other two at the reference values.
# (name, index of the varied parameter, colour, line style, marker, axis)
sweeps = [
    ("p_pueR2", 0, "tab:blue", "-", "o"),
    ("p_pueR3", 1, "tab:orange", "--", "s"),
    ("p_max", 2, "tab:green", ":", "^"),
]


def sweep_members(i_var):
    """Indices of runs where only parameter i_var differs from the reference."""
    idx = [i for i, r in enumerate(results)
           if all(close(r[k], results[i_ref][k]) for k in range(3) if k != i_var)]
    return sorted(idx, key=lambda i: results[i][i_var])


fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(17, 4.8),
                                    gridspec_kw={"width_ratios": [1, 0.8, 1.3]})
in_sweep = set()
for name, i_var, color, style, marker in sweeps:
    idx = sweep_members(i_var)
    if len(idx) < 2:
        continue
    in_sweep.update(idx)
    ax = ax2 if i_var == 2 else ax1
    xs = [results[i][i_var] for i in idx]
    other = ", ".join("{} = {:g}".format(n, results[i_ref][k])
                      for n, k, *_ in sweeps if k != i_var)
    ax.plot(xs, means[idx], style, marker=marker, color=color,
            label="{} varied ({})".format(name, other))
    for i in idx:
        if i != i_ref:
            ax.annotate("{:+.1f}%".format(change[i]), (results[i][i_var], means[i]),
                        textcoords="offset points", xytext=(0, 8), ha="center",
                        fontsize=8, color=color)
    # (c) daily values: darker = further from the case value
    shades = np.linspace(1.0, 0.45, len(idx))
    order_idx = idx[::-1] if i_var < 2 else idx
    for i, shade in zip(order_idx, shades):
        if i == i_ref:
            continue
        rgb = np.array(plt.matplotlib.colors.to_rgb(color))
        ax3.plot(results[i][3], results[i][4], style, color=rgb * shade, lw=1.2,
                 label="{} = {:g}".format(name, results[i][i_var]))
ax3.plot(results[i_ref][3], results[i_ref][4], "-", color="black", lw=1.4,
         label="reference ({:g}/{:g}, p_max {:g})".format(ref_r2, ref_r3, ref_pmax))
for i, r in enumerate(results):                     # runs with more than one change
    if i not in in_sweep:
        lab = "R2 {:g}, R3 {:g}, p_max {:g}".format(*r[:3])
        ax1.scatter([r[0]], [means[i]], color="tab:purple", marker="D", zorder=3, label=lab)
        ax1.annotate("{:+.1f}%".format(change[i]), (r[0], means[i]), textcoords="offset points",
                     xytext=(0, -14), ha="center", fontsize=8, color="tab:purple")
        ax3.plot(r[3], r[4], "-.", color="tab:purple", label=lab)

for ax in (ax1, ax2):
    ax.axhline(ref, color="0.7", lw=0.8, zorder=0)
    ax.set_ylabel("annual mean {} [{}]".format(var, units))
    ax.margins(y=0.12)
    if ax.lines[1:] or ax.collections:
        ax.legend(fontsize=8, frameon=False)
ax1.set_xlabel("excreted fraction of TEP uptake")
ax1.set_title("(a) {}, mean over {}".format(long_name, year), fontsize=10)
ax1.invert_xaxis()                                  # from 0.90 (case) to 0.10
ax2.set_xscale("log")
pmax_values = sorted({r[2] for i, r in enumerate(results) if i in sweep_members(2)})
if pmax_values:
    ax2.set_xticks(pmax_values)
    ax2.set_xticklabels(["{:g}".format(v) for v in pmax_values])
    ax2.minorticks_off()
ax2.set_xlabel("p_max (extraction cap / R2 uptake factor)")
ax2.set_title("(b) p_max, mean over {}".format(year), fontsize=10)
ax2.axvline(ref_pmax, color="0.7", lw=0.8, ls="--", zorder=0)

ax3.set_ylabel("{} [{}]".format(var, units))
ax3.set_title("(c) daily values in {}".format(year), fontsize=10)
ax3.legend(fontsize=7, frameon=False, loc="upper left", bbox_to_anchor=(1.01, 1.0))
fig.autofmt_xdate()
for ax in (ax1, ax2):
    plt.setp(ax.get_xticklabels(), rotation=0, ha="center")
fig.tight_layout()

# Save to Cloud Storage: /tmp/data is cleared when the workflow finishes
results_folder = str(param_cloud_folder)
# Empty the Cloud Storage folder first, so files from an earlier submission are not mixed in
if os.path.isdir(results_folder) and os.listdir(results_folder):
    print("Emptying", results_folder)
    for name in os.listdir(results_folder):
        remove_tree(os.path.join(results_folder, name))
os.makedirs(results_folder, exist_ok=True)
figure_file = os.path.join(results_folder, "TEPexcretion_{}_{}.png".format(var, year))
fig.savefig(figure_file, dpi=150)
with open(os.path.join(results_folder, "TEPexcretion_{}_{}.csv".format(var, year)), "w") as f:
    f.write("\n".join(table) + "\n")
for run_dir in folders:                  # keep the model logs and run info as well
    name = os.path.basename(os.path.normpath(run_dir))
    for fname in ("log.out", "run_info.json"):
        if os.path.exists(os.path.join(run_dir, fname)):
            os.makedirs(os.path.join(results_folder, name), exist_ok=True)
            shutil.copy(os.path.join(run_dir, fname), os.path.join(results_folder, name, fname))
print("results_folder:", results_folder)
print("figure:", figure_file)
plt.show()
````
